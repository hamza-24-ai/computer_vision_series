```python
import cv2
import numpy as np
```

```python
org_img = cv2.imread("06.jpeg")

org = cv2.resize(org_img,(300,500))

g = cv2.GaussianBlur(org,(7,7),0)
m = cv2.medianBlur(org,(5))
b = cv2.bilateralFilter(org,10,100,100)

h = np.hstack((org,g,m,b))

cv2.imshow("hamza",h)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
