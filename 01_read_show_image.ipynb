```python
import cv2
import numpy as np
```

```python
img = cv2.imread("smart.png")
cv2.imshow("Muhammad_Hamza", img)
cv2.waitKey(0)
# cv2.destroyWindow() well it required number
cv2.destroyAllWindows()
```

# Show Multiple Images in one frame

```python
img = cv2.imread("smart.png")
re_img = cv2.resize(img,(450,250))

h = np.hstack((re_img,re_img))
v = np.vstack((h,h))

cv2.imshow("Hamza",v)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

#  Show image by press button

```python
array = ["01.png","02.png","03.png","04.png","05.png"]
print(array)
```

```python
for im in array:
    img = cv2.imread(im)
    re_img = cv2.resize(img,(1050,650))
    cv2.imshow("Hamza",re_img)
    cv2.waitKey(0)

cv2.destroyAllWindows()
```
