```python
import cv2
```

```python
img = cv2.imread("smartLab.png")
img = cv2.resize(img,(600,400))

new_img = cv2.Canny(img, 50,50, apertureSize=5, L2gradient=True)
new_img1 = cv2.Canny(img, 50,50,)

# with Edges
cv2.imshow("hamza2",new_img1)

# with detail Edges
cv2.imshow("hamza1",new_img)
# Without Edges
cv2.imshow("hamza",img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
