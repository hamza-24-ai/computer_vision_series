```python
import cv2
```

```python

img = cv2.imread("01.png")
img = cv2.resize(img,(800,600))

# Enter font personalities

txt = cv2.putText(
    img = img,
    text = "Muhammad Hamza",
    org = (50,80),
    fontFace = cv2.FONT_HERSHEY_DUPLEX,
    fontScale = 2,
    color = (0,0,255),
    thickness = 3,
    lineType = cv2.LINE_8,
    bottomLeftOrigin = False
)


txt_2 = cv2.putText(
    img = txt,
    text = "Muhammad Hamza",
    org = (50,90),
    fontFace = cv2.FONT_HERSHEY_DUPLEX,
    fontScale = 2,
    color = (255,0,255),
    thickness = 3,
    lineType = cv2.LINE_8,
    bottomLeftOrigin = True
)

cv2.imshow("hamza",txt)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python

img1 = cv2.imread("smartLab.png")
img1 = cv2.resize(img1,(800,600))

# Make a rectangle shape

cv2.rectangle(
    img = img1,
    pt1 = (220,310),
    pt2 = (340,440),
    color = (0,255,0),
    thickness = 4,
    lineType = 2
)


cv2.imshow("hamza",img1)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
